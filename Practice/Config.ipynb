Be careful When copying array !!!!

```python
b = np.array([1,2,3])
c = b
c  
```

```python
print(b)
print(c)
b[0] = 100
print(b)
print(c)
```

d = b.copy()

```python
d = b.copy()
d[0] = 1
print(d) 
print(b)
```
