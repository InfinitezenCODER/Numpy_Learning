```python
import numpy as np
```

```python
V1 = np.array([1,2,3])
V2 = np.array([5, 6, 7])
A = np.array([[1,2],[3,4]])
B = np.array([[6,7],[8,9]])
```

Matrix Multiplication :
Use @ or matmul() for matrix multiplication .

```python
C = A @ B
C = np.matmul(A,B)
print(C)
```

```python

# Dot Product :
# Use np.dot() .
dot = np.dot(V1,V2)
print(dot)
```

```python
# Deteminant
I = np.identity(3)
print(np.linalg.det(I))
```

```python
AT = A.T
print(AT)
```

```python
# Inverse Matrix
A_inv = np.linalg.inv(A)
print(A_inv)
```

Advanced Linear Algebra  

```python
np.linalg.matrix_rank(A)
```

```python
# norm = It measures the Length of the vector 
v = np.array([3,4])
np.linalg.norm(v)
```

```python
eigenvalues, eigenvectors = np.linalg.eig(A)

print(eigenvalues)
print(eigenvectors)
```
