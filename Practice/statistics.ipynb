```python
np.sum(stats, axis=0)
np.sum(stats, axis=1)
```

Vertical Stacking Vectors 

```python
VS1 = np.array([1,2,3])
VS2 = np.array([4,5,6])
result = np.vstack((VS1,VS2,VS1))
result 
```

Horizontal Stack 

```python
HS1 = np.ones([3,3])
HS2 = np.zeros([3,4])
print (HS1)
print (HS2)
```

```python
Result = np.hstack((HS1,HS2))
Result
```
