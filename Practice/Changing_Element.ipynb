Accesinng/Changing specific element .


```python
import numpy as np
```

```python
a = np.array([[1, 2, 3, 4, 5,6,8,5],[6,7,8,9,10,11,12,13]])
print(a)
```

```python
a.shape
```

```python
#Get specifice element. [r,c]
a[1,4]
```

```python
# Negative Indexing . Last column will be -1
a[1,-2]
```

```python
# Get a specific row . [r, :]]
a[1, :]
```

```python
# Get a specific column . [:, c]]
a[:, 3]
```

```python
# Sllicing of array . [start:stop:jump]
# Start from index : stop before this index : jump
a[1, 1:4:]
```

```python
a[0, 1:4:]
```

```python
a[1, 1:-6:2]
```

```python
a[1,1:-1:2]
```

Altering Data


```python
a[0,7] = 10
print(a)
```

```python
a[:,-6] = 69
print(a)
```

```python
a[:,-7]= [6,9]
print(a)
```
